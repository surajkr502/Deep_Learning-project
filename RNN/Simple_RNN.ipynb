```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
```

```python
model = Sequential([SimpleRNN(32, input_shape=(10, 5)),
    Dense(1, activation="sigmoid")
])
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.summary()
```
